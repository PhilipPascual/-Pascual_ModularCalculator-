```python

def add_numbers(number_1, number_2):
    return number_1 + number_2

def subtract_numbers(number_1, number_2):
    return number_1 - number_2

def multiply_numbers(number_1, number_2):
    return number_1 * number_2

def divide_numbers(number_1, number_2):
    quotient = number_1 / number_2
    return quotient



number_1 = float(input("Enter first number: "))
number_2 = float(input("Enter second number: "))

print("Choose: 1 as Addition, 2 as Subtraction, 3 as Multiplication, and 4 as Division")
option = int(input("Enter option: "))


if option == 1:
    print(f"Sum is: {add_numbers(number_1, number_2)}")
elif option == 2:
    print(f"Difference is: {subtract_numbers(number_1, number_2)}")
elif option == 3:
    print(f"Product is: {multiply_numbers(number_1, number_2)}")
elif option == 4:
    print(f"Quotient is: {round(divide_numbers(number_1, number_2), 2)}")
else:
    print("Invalid Input")

# 1. The functions I used was user defined functions and pre_Defined functions
# 2. I only use 2 parameters
# 3. The arguments used was the user input of both variable num1 and num2
# 4. The program used the returned value as a way to compute it
# 5. It is better to use functions because it is easy to be debugged and not complicated compared to one single block of code
```
