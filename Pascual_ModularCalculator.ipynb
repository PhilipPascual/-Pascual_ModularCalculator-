```python
def add_numbers(num1, num2):
  return num1 + num2


def subtract_numbers(num1, num2):
  return num1 - num2


def multiply_numbers(num1, num2):
  return num1 * num2


def divide_numbers(num1, num2):
  if num2 == 0:
    return "Invalid Input"
  else:
    return num1 / num2


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print(
    "Choose: 1 as Addition, 2 as Subtraction, 3 as Multiplication, and 4 as"
    " Division"
)
option = int(input("Enter option: "))

if option == 1:
  print(f"Sum is: {add_numbers(num1, num2)}")
elif option == 2:
  print(f"Difference is: {subtract_numbers(num1, num2)}")
elif option == 3:
  print(f"Product is: {multiply_numbers(num1, num2)}")
elif option == 4:
  result = divide_numbers(num1, num2)
  if result == "Invalid Input":
    print(result)
  else:
    print(f"Quotient is: {round(result, 2)}")
else:
    print("Invalid Input")

  # 1. The functions I used are Pre_defined functions and User Defined Functions
  # 2. The parameters I used are num1 aand num2
  # 3. The arguments passed when the functions were called are num1 and num2
  # 4. The program used the returned values as a way to store it
  # 5. It is better to divide the program into seperate functions because it is easier and also quick to debug
```
